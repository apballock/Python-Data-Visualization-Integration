# Python & MySQL Integration
## Connecting Python to MySQL
mysql-connector-python = the bridge between Python and MySQL Workbench
pandas.read_sql() = run a SQL query and load results directly into a DataFrame

```python
print("hello")
```

```python
import mysql.connector
import pandas as pd

conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="12345678",
    database="transactions"
)

print("Connection status:", conn.is_connected())
```

```python
# Task 2 — Criar cursor e listar tabelas
cursor = conn.cursor()
cursor.execute("SHOW TABLES;")

print("Tables in database:")
for table in cursor.fetchall():
    print(" -", table[0])
```

```python
# Task 3 — Carregar tabela transaction como DataFrame
query = "SELECT * FROM transaction LIMIT 100;"
df_mysql = pd.read_sql(query, conn)

print("Shape:", df_mysql.shape)
print(df_mysql.head())
```

```python
# Task 4 — Query com agregação
query = """
    SELECT 
        company_id,
        COUNT(*) as total_transactions,
        ROUND(SUM(amount), 2) as total_amount,
        ROUND(AVG(amount), 2) as avg_amount,
        SUM(declined) as total_declined
    FROM transaction
    GROUP BY company_id
    ORDER BY total_amount DESC
    LIMIT 10;
"""
df_summary = pd.read_sql(query, conn)
print(df_summary)
```

```python
# Task 5 — Fechar conexão
cursor.close()
conn.close()
print("Connection closed.")
```
