# Power BI & Python Integration
## Running Python Scripts in Power BI
Power BI can execute Python scripts to transform data and create visuals.
Python visuals in Power BI use Matplotlib or Seaborn under the hood.

```python
# Exercise 1 — Python Script in Power BI (Data Transform)
# Step 1: Open Power BI Desktop → Get Data → Python Script
# Step 2: Paste the script below in Power BI's Python editor

import pandas as pd

df = pd.read_csv("financials_clean2.csv")
summary = df.groupby('Segment').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Avg_Margin=('Profit Margin %', 'mean'),
    Count=('Sales', 'count')
).round(2).reset_index()
```
